```python
def token(txt):
    token = nltk.word_tokenize(txt)
    a = [stemmer.stem(w) for w in token]
    return " ".join(a)
```

Recommender Function

```python
def recommendation(song_name):
    result = df[df['song'].str.lower() == song_name.lower()]
    
    if result.empty:
        print("Song not found")
        return []
    
    idx = result.index[0]
    
    distances = sorted(
        list(enumerate(similar[idx])),
        reverse=True,
        key=lambda x: x[1]
    )
    
    songs = []
    for s_id in distances[1:5]:
        songs.append(df.iloc[s_id[0]].song)
    
    return songs
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
hindi_df    = pd.read_csv("Hindi_songs.csv")
punjabi_df  = pd.read_csv("Punjabi_songs.csv")
haryanvi_df = pd.read_csv("Haryanvi_songs.csv")

new_songs = pd.concat([hindi_df, punjabi_df, haryanvi_df], ignore_index=True)
```

```python
def features_to_text(row):
    words = []
    if row['danceability'] > 0.7:  words += ['dance', 'dance', 'groove']
    if row['energy'] > 0.7:        words += ['energy', 'energy', 'hype']
    if row['acousticness'] > 0.5:  words += ['acoustic', 'soft', 'mellow']
    if row['liveness'] > 0.7:      words += ['live', 'concert', 'crowd']
    if row['Valence'] > 0.7:       words += ['happy', 'happy', 'upbeat']
    if row['Valence'] < 0.3:       words += ['sad', 'sad', 'emotional']
    if row['tempo'] > 120:         words += ['fast', 'beat', 'rhythm']
    if row['speechiness'] > 0.3:   words += ['rap', 'rap', 'words']
    words += [row['language'].lower().replace(' ', '_')]
    return ' '.join(words) if words else 'music song'

new_songs['text'] = new_songs.apply(features_to_text, axis=1)
```

```python
new_songs = new_songs.rename(columns={
    'song_name': 'song',
    'singer':    'artist'
})[['song', 'artist', 'text']]
```

```python
df = pd.concat([df, new_songs], ignore_index=True).drop_duplicates(subset='song').reset_index(drop=True)
print(f"Total songs now: {len(df)}")
print(df.tail(5))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

tfid   = TfidfVectorizer(analyzer='word', stop_words='english')
matrix = tfid.fit_transform(df['text'])
similar = cosine_similarity(matrix)
```

```python
import pickle
pickle.dump(df,      open("df",     "wb"))
pickle.dump(similar, open("similar", "wb"))
print("✅ Done! New df and similar files saved.")
```
